```python
import pandas as pd
import numpy as np
import os

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

DATA_PATH = "../data/raw"

print("Project started successfully!")
```

```python
print("Files available in raw data folder:")
print(os.listdir(DATA_PATH))
```

```python
attrition_df = pd.read_csv(
    f"{DATA_PATH}/employee_attrition.csv"
)

engagement_df = pd.read_csv(
    f"{DATA_PATH}/Cleaned_HR_Data_Analysis.csv"
)

occupation_df = pd.read_csv(
    f"{DATA_PATH}/occupation_data.csv"
)

essential_skills_df = pd.read_csv(
    f"{DATA_PATH}/essential_skills.csv"
)

software_skills_df = pd.read_csv(
    f"{DATA_PATH}/software_skills.csv"
)
```

```python
datasets = {
    "Attrition": attrition_df,
    "Engagement": engagement_df,
    "Occupation": occupation_df,
    "Essential Skills": essential_skills_df,
    "Software Skills": software_skills_df
}

for name, df in datasets.items():
    print("=" * 60)
    print(name)
    print("=" * 60)
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
    print()
```

```python
for name, df in datasets.items():
    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    for i, column in enumerate(df.columns, start=1):
        print(f"{i}. {column}")
```

```python
for name, df in datasets.items():
    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    display(
        pd.DataFrame({
            "column": df.columns,
            "dtype": df.dtypes.astype(str).values,
            "unique_values": [
                df[col].nunique() for col in df.columns
            ]
        })
    )
```

```python
for name, df in datasets.items():

    missing = (
        df.isnull()
        .sum()
        .sort_values(ascending=False)
    )

    missing = missing[missing > 0]

    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    if missing.empty:
        print("No missing values found.")
    else:
        display(missing)
```

```python
for name, df in datasets.items():

    print(
        f"{name}: {df.duplicated().sum()} duplicate rows"
    )
```

```python
print("Attrition distribution:")
display(attrition_df["Attrition"].value_counts())

print("\nAttrition percentage:")
display(
    attrition_df["Attrition"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
categorical_columns = attrition_df.select_dtypes(
    include="object"
).columns

for col in categorical_columns:
    print("\n", "=" * 50)
    print(col)
    print("=" * 50)
    print(attrition_df[col].value_counts(dropna=False))
```

```python
numeric_columns = attrition_df.select_dtypes(
    include=np.number
).columns

display(
    attrition_df[numeric_columns].describe().T
)
```

```python
print("Attrition employee ID candidates:")
print(
    attrition_df[
        ["EmployeeNumber"]
    ].head(10)
)
```

```python
print(
    "Unique EmployeeNumber:",
    attrition_df["EmployeeNumber"].nunique()
)

print(
    "Total rows:",
    len(attrition_df)
)
```

```python
print("Attrition ID columns:")
print(
    [
        c for c in attrition_df.columns
        if "id" in c.lower() or "number" in c.lower()
    ]
)

print("\nEngagement ID columns:")
print(
    [
        c for c in engagement_df.columns
        if "id" in c.lower() or "number" in c.lower()
    ]
)
```

```python
attrition_roles = set(
    attrition_df["JobRole"]
    .dropna()
    .astype(str)
    .str.strip()
    .str.lower()
    .unique()
)

occupation_titles = set(
    occupation_df["Title"]
    .dropna()
    .astype(str)
    .str.strip()
    .str.lower()
    .unique()
)

role_matches = attrition_roles.intersection(
    occupation_titles
)

print("Attrition roles:", len(attrition_roles))
print("Occupation titles:", len(occupation_titles))
print("Exact normalized matches:", len(role_matches))

print("\nMatches:")
for role in sorted(role_matches):
    print(role)
```

```python
profile = []

for name, df in datasets.items():

    profile.append({
        "dataset": name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "duplicate_rows": df.duplicated().sum(),
        "missing_cells": int(df.isnull().sum().sum()),
        "memory_mb": round(
            df.memory_usage(deep=True).sum() / 1024**2,
            2
        )
    })

profile_df = pd.DataFrame(profile)

display(profile_df)
```

```python
profile_df.to_csv(
    "../data/processed/dataset_profile.csv",
    index=False
)
```
