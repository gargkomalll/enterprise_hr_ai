# Step 01: Data Understanding & Initial Profiling

## Overview
This notebook performs exploratory profiling across the 5 core CSV datasets for the **Enterprise HR AI** platform:
1. `employee_attrition.csv`
2. `hr_performance_engagement.csv`
3. `occupation_data.csv`
4. `essential_skills.csv`
5. `software_skills.csv`

For each dataset, we assess shape, data types, missing value percentages, duplicate counts, primary key candidates, target class balance (where applicable), and join relationships.


```python
import pandas as pd
import numpy as np
import os

RAW_DIR = os.path.join("..", "data", "raw")
print("Raw Data Directory:", os.path.abspath(RAW_DIR))
```

---
## 1. Employee Attrition Dataset (`employee_attrition.csv`)
Target dataset containing employee demographic, job, and attrition status.


```python
attr_df = pd.read_csv(os.path.join(RAW_DIR, "employee_attrition.csv"))
print(f"Shape: {attr_df.shape} (Rows: {attr_df.shape[0]}, Cols: {attr_df.shape[1]})")
print("\nColumns & Dtypes:")
print(attr_df.dtypes)

print("\nMissing Values:")
missing = attr_df.isnull().sum()
print(missing[missing > 0] if missing.sum() > 0 else "No missing values.")

print(f"\nDuplicate rows: {attr_df.duplicated().sum()}")
print(f"Unique EmployeeNumbers: {attr_df['EmployeeNumber'].nunique()} / {len(attr_df)}")

print("\n--- Target Class Balance ('Attrition') ---")
target_counts = attr_df['Attrition'].value_counts(dropna=False)
target_props = attr_df['Attrition'].value_counts(normalize=True, dropna=False) * 100
target_summary = pd.DataFrame({'Count': target_counts, 'Percentage (%)': target_props})
print(target_summary)
```

---
## 2. HR Performance & Engagement Dataset (`hr_performance_engagement.csv`)
Contains employee performance ratings, survey engagement, satisfaction, and work-life balance scores.


```python
perf_df = pd.read_csv(os.path.join(RAW_DIR, "hr_performance_engagement.csv"))
print(f"Shape: {perf_df.shape} (Rows: {perf_df.shape[0]}, Cols: {perf_df.shape[1]})")
print("\nColumns & Dtypes:")
print(perf_df.dtypes)

print("\nMissing Values:")
missing = perf_df.isnull().sum()
print(missing[missing > 0] if missing.sum() > 0 else "No missing values.")

print(f"\nDuplicate rows: {perf_df.duplicated().sum()}")
print(f"Unique Employee IDs: {perf_df['Employee ID'].nunique()} / {len(perf_df)}")

print("\nEngagement & Satisfaction Score Distributions:")
print(perf_df[['Engagement Score', 'Satisfaction Score', 'Work-Life Balance Score', 'Current Employee Rating']].describe())
```

---
## 3. Occupation Data Master (`occupation_data.csv`)
O*NET master reference table mapping SOC job codes to standardized titles and descriptions.


```python
occ_df = pd.read_csv(os.path.join(RAW_DIR, "occupation_data.csv"))
print(f"Shape: {occ_df.shape} (Rows: {occ_df.shape[0]}, Cols: {occ_df.shape[1]})")
print("\nColumns & Dtypes:")
print(occ_df.dtypes)

print("\nMissing Values:")
missing = occ_df.isnull().sum()
print(missing[missing > 0] if missing.sum() > 0 else "No missing values.")

print(f"\nDuplicate rows: {occ_df.duplicated().sum()}")
print(f"Unique O*NET-SOC Codes: {occ_df['O*NET-SOC Code'].nunique()} / {len(occ_df)}")
```

---
## 4. Essential Skills Dataset (`essential_skills.csv`)
Requirements table mapping O*NET-SOC codes to essential role skills and importance ratings.


```python
ess_df = pd.read_csv(os.path.join(RAW_DIR, "essential_skills.csv"))
print(f"Shape: {ess_df.shape} (Rows: {ess_df.shape[0]}, Cols: {ess_df.shape[1]})")
print("\nColumns & Dtypes:")
print(ess_df.dtypes)

print("\nMissing Values:")
missing = ess_df.isnull().sum()
print(missing[missing > 0])

print(f"\nDuplicate rows: {ess_df.duplicated().sum()}")
print(f"Unique O*NET-SOC Codes: {ess_df['O*NET-SOC Code'].nunique()}")
print(f"Unique Skill Names ('Element Name'): {ess_df['Element Name'].nunique()}")
```

---
## 5. Software Skills Dataset (`software_skills.csv`)
Specific software tools, programming languages, and technologies per O*NET role.


```python
soft_df = pd.read_csv(os.path.join(RAW_DIR, "software_skills.csv"))
print(f"Shape: {soft_df.shape} (Rows: {soft_df.shape[0]}, Cols: {soft_df.shape[1]})")
print("\nColumns & Dtypes:")
print(soft_df.dtypes)

print("\nMissing Values:")
missing = soft_df.isnull().sum()
print(missing[missing > 0] if missing.sum() > 0 else "No missing values.")

print(f"\nDuplicate rows: {soft_df.duplicated().sum()}")
print(f"Unique O*NET-SOC Codes: {soft_df['O*NET-SOC Code'].nunique()}")
print(f"Unique Software Tools ('Workplace Example'): {soft_df['Workplace Example'].nunique()}")
```
