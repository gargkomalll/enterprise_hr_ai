# Step 11: Role Intelligence Master Table

## Overview
This notebook builds the role-master reference table linking internal enterprise HR job roles to standardized O*NET SOC occupation taxonomy:
- Maps all 9 internal HR job roles to O*NET SOC codes.
- Joins official SOC titles and descriptions from `occupation_master.csv`.
- Exports `data/processed/role_intelligence_master.csv`.


```python
import pandas as pd
import numpy as np
import os

PROCESSED_DIR = os.path.join("..", "data", "processed")

occ_df = pd.read_csv(os.path.join(PROCESSED_DIR, "occupation_master.csv"))
attr_df = pd.read_csv(os.path.join(PROCESSED_DIR, "employee_attrition_processed.csv"))

print(f"Occupation Master shape: {occ_df.shape}")
print(f"Unique internal HR JobRoles: {attr_df['JobRole'].unique()}")
```

---
## 1. Map HR Job Roles to O*NET SOC Standard


```python
role_mapping = [
    {'HR_Job_Role': 'Sales Executive', 'ONET_SOC_Code': '41-4012.00'},
    {'HR_Job_Role': 'Research Scientist', 'ONET_SOC_Code': '19-1029.01'},
    {'HR_Job_Role': 'Laboratory Technician', 'ONET_SOC_Code': '29-2012.00'},
    {'HR_Job_Role': 'Manufacturing Director', 'ONET_SOC_Code': '11-1021.00'},
    {'HR_Job_Role': 'Healthcare Representative', 'ONET_SOC_Code': '41-4011.00'},
    {'HR_Job_Role': 'Manager', 'ONET_SOC_Code': '11-1021.00'},
    {'HR_Job_Role': 'Sales Representative', 'ONET_SOC_Code': '41-4012.00'},
    {'HR_Job_Role': 'Research Director', 'ONET_SOC_Code': '11-9121.00'},
    {'HR_Job_Role': 'Human Resources', 'ONET_SOC_Code': '13-1071.00'}
]

role_map_df = pd.DataFrame(role_mapping)

# Join with O*NET Occupation Master
role_master = pd.merge(role_map_df, occ_df, left_on='ONET_SOC_Code', right_on='O*NET-SOC Code', how='left')
role_master.drop(columns=['O*NET-SOC Code'], inplace=True)
role_master.rename(columns={'Title': 'ONET_Title', 'Description': 'ONET_Description'}, inplace=True)

print("=== Enterprise Role Intelligence Master Table ===")
print(role_master[['HR_Job_Role', 'ONET_SOC_Code', 'ONET_Title']].to_string(index=False))

out_path = os.path.join(PROCESSED_DIR, "role_intelligence_master.csv")
role_master.to_csv(out_path, index=False)
print(f"\nSaved Role Master: {out_path}")
```
