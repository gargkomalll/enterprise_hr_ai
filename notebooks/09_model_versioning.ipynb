# Step 09: Model Versioning & Registry Packaging

## Overview
This notebook packages the winning model under a versioned model registry path (`models/v1/`):
- Saves pipeline to `models/v1/model.joblib`.
- Generates structured metadata schema `models/v1/metadata.json`.
- Documents future integration point for MLflow / Evidently AI model registries.


```python
import pandas as pd
import numpy as np
import os
import joblib
import json
from datetime import datetime, timezone

PROCESSED_DIR = os.path.join("..", "data", "processed")
MODELS_DIR = os.path.join("..", "models")
V1_DIR = os.path.join(MODELS_DIR, "v1")
os.makedirs(V1_DIR, exist_ok=True)

pipeline = joblib.load(os.path.join(MODELS_DIR, "attrition_pipeline.joblib"))

# Save model artifact to v1 registry
v1_model_path = os.path.join(V1_DIR, "model.joblib")
joblib.dump(pipeline, v1_model_path)

metadata = {
    "model_name": "enterprise_hr_attrition_classifier",
    "version": "v1.0.0",
    "algorithm": "LogisticRegression(class_weight='balanced', max_iter=1000)",
    "training_date": datetime.now(timezone.utc).isoformat(),
    "framework": "scikit-learn",
    "target": "Attrition",
    "metrics": {
        "precision": 0.4286,
        "recall": 0.7660,
        "f1_score": 0.5496,
        "roc_auc": 0.8258
    },
    "features_count": len(pipeline.named_steps['preprocessor'].get_feature_names_out()),
    "mlflow_integration_note": "MLflow tracking can be enabled here using mlflow.sklearn.log_model() and mlflow.log_metrics()."
}

v1_meta_path = os.path.join(V1_DIR, "metadata.json")
with open(v1_meta_path, 'w') as f:
    json.dump(metadata, f, indent=4)

print(f"✔ Saved Model v1 Artifact: {v1_model_path}")
print(f"✔ Saved Metadata File: {v1_meta_path}")
print("\nMetadata Content:")
print(json.dumps(metadata, indent=2))
```
